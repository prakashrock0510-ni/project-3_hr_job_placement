```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import(accuracy_score,classification_report,confusion_matrix)
```

```python
df=pd.read_csv("/content/Job_placement_cleaneddata.csv")
```

```python
df
```

```python
X=df.drop(columns=["status"])
y=df["status"]
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=42,stratify=y)
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

LOGISTIC REGRESSION

```python
logistic_model = LogisticRegression(max_iter=3000)
logistic_model.fit(X_train, y_train)
logistic_predictions = logistic_model.predict(X_test)
print("Logistic Regression completed")
```

```python
logistic_accuracy = accuracy_score(y_test,logistic_predictions)
print("Logistic Regression Accuracy:", round(logistic_accuracy, 4))
print("Accuracy Percentage:", round(logistic_accuracy * 100, 2), "%")
```

```python
print("Logistic Regression Classification Report")
print(classification_report(y_test,logistic_predictions,labels=[0, 1],target_names=["Not Placed", "Placed"],zero_division=0))
```

```python
logistic_cm = confusion_matrix(y_test,logistic_predictions,labels=[0, 1])
print("Logistic Regression Confusion Matrix")
print(logistic_cm)
```

2.Decision Tree

```python
decision_tree_model = DecisionTreeClassifier(max_depth=10,random_state=42)
decision_tree_model.fit(X_train, y_train)
tree_predictions = decision_tree_model.predict(X_test)
print("Decision Tree completed")
```

```python
tree_accuracy = accuracy_score(y_test,tree_predictions)
print("Decision Tree Accuracy:", round(tree_accuracy, 4))
print("Accuracy Percentage:", round(tree_accuracy * 100, 2), "%")
```

```python
print("Decison Tree report")
print(classification_report(y_test,tree_predictions,labels=[0, 1],target_names=["Not Placed", "Placed"],zero_division=0))
```

```python
tree_cm = confusion_matrix(y_test,tree_predictions,labels=[0, 1])
print("Decision Tree Confusion Matrix")
print(tree_cm)
```

3.Random Forest


```python
random_forest_model = RandomForestClassifier(n_estimators=100,max_depth=15,random_state=42)
random_forest_model.fit(X_train, y_train)
forest_predictions = random_forest_model.predict(X_test)
print("Random Forest completed")
```

```python
forest_accuracy = accuracy_score(y_test,forest_predictions)
print("Random Forest Accuracy:", round(forest_accuracy, 4))
print("Accuracy Percentage:", round(forest_accuracy * 100, 2), "%")
```

```python
print("Random Forest Classification Report")
print(classification_report(y_test,forest_predictions,labels=[0, 1],target_names=["Not Placed", "Placed"],zero_division=0))
```

```python
forest_cm = confusion_matrix(y_test,forest_predictions,labels=[0, 1])
print("Random Forest Confusion Matrix")
print(forest_cm)
```

Model Comparison

```python
model_comparison = pd.DataFrame({"Model":["Logistic Regression","Decision Tree","Random Forest"],
    "Accuracy": [logistic_accuracy,tree_accuracy,forest_accuracy]})
model_comparison["Accuracy Percentage"] =(model_comparison["Accuracy"] * 100).round(2)
model_comparison = model_comparison.sort_values(by="Accuracy",ascending=False).reset_index(drop=True)
print(model_comparison)
```

```python
feature_importance = pd.DataFrame({"Feature": X_train.columns,"Importance": random_forest_model.feature_importances_})
feature_importance = feature_importance.sort_values(by="Importance",ascending=False)
print(feature_importance.head(15))
```

```python
placement_probabilities = random_forest_model.predict_proba(X_test)
placed_index = list(random_forest_model.classes_).index(1)
placed_probability = placement_probabilities[:, placed_index]
probability_results = pd.DataFrame({"Actual_Status": y_test.to_numpy()[:10],"Placement_Probability": placed_probability[:10]})
probability_results["Placement_Probability"] = (probability_results["Placement_Probability"] * 100).round(2)
print(probability_results)
```
